### 사용자 답변에 대한 키워드 추출

```python
# Java 서버의 키워드 추출 요청 더미 객체
keyword_extraction_request = {
    "user_input": "바다가 유명한 지역을 여행하고 싶어",
    "prompt": """
    다음 사용자 입력에서 여행 계획과 관련된 핵심 키워드들을 추출해주세요.
        
    사용자 입력: {user_input}
        
    추출 기준:
    1. 지역/도시명 (예: 부산, 제주도, 서울)
    2. 여행 활동 (예: 해수욕, 등산, 쇼핑)
    3. 숙박 유형 (예: 호텔, 펜션, 게스트하우스)
    4. 음식/맛집 (예: 해산물, 카페, 맛집)
    5. 관광지 유형 (예: 해변, 산, 박물관, 사찰, 바다)
    6. 분위기/특성 (예: 로맨틱, 가족여행, 힐링)

    결과는 쉼표로 구분된 키워드 목록으로 반환해주세요.
    예시: 부산, 해변, 카페, 맛집, 로맨틱
    """
}
```

### 키워드 추출 함수 정의

```python
import os
from typing import List
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

os.environ["OPENAI_API_KEY"] = "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"

async def extract_keywords(user_input: str, prompt: str) -> List[str]:
    # 키워드 추출 로직 구현

    # 입력 매개변수 확인
    print("사용자 입력 : ", user_input)
    print("사용할 프롬프트", prompt)

    # RDB에 저장된 프롬프트(단순한 문자열) 템플릿을 객체 형태로 변환
    template = ChatPromptTemplate.from_template(prompt)

    # 사용자의 입력을 템플릿에 넣어서 LLM에 전달할 메세지 만들기
    # 왼쪽 user_input은 {user_input}에 채워질 키
    # 오른쪽은 사용자가 입력한 값
    message = template.format_messages(user_input = user_input)

    # llm 설정, 모델 및 temperature 설정. temperature가 낮을수록 창의성이 떨어짐
    llm = ChatOpenAI(
        model = "gpt-4o-mini",
        temperature = 0.5
    )

    # llm에게 명령을 전달하고 답변을 받는다. (ainvoke의 역할)
    response = await llm.ainvoke(message)

    print("사용자 입력 : ", message[0].content)
    
    # 답변 내용 확인하기
    print(response.content)

    # list 형태로 분할해서 반환시킨다.
    return response.content.split(", ")

user_input = keyword_extraction_request["user_input"]
prompt = keyword_extraction_request["prompt"]
extracted_keywords = await extract_keywords(user_input, prompt)
print(user_input)
print(prompt)
print(extracted_keywords)
```

### 임베딩 및 벡터 DB에서 PK 추출

```python
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_core.documents import Document

# 벡터DB 스키마 생성
if not os.path.exists("./chroma_region_test"):
    os.makedirs("./chroma_region_test")

# 벡터DB 설정
vector_store = Chroma(
    collection_name = "region_collection",
    embedding_function = OpenAIEmbeddings(
        model = "text-embedding-3-large"
    ),
    persist_directory = "./chroma_region_test"
)

# 더미 데이터를 임베딩
description = "dumy_region 은 해변, 바다가 유명하고 해산물이 맛있는 지역입니다."
metadata = {"pk" : "dumy_pk_001"}
document = Document(
    page_content = description,
    metadata = metadata
)


# print(document.page_content)
documents = []
documents.append(document)
vector_store.add_documents(documents)

# 유사도 검색에 사용할 쿼리 생성
query = " ".join(extracted_keywords)

# 유사도 검색
search_result = vector_store.similarity_search(query)

# 결과에서 PK 추출
pks = []
for doc in search_result:
    pk = doc.metadata.get("pk")
    pks.append(pk)

# 결과 PK 리스트 확인
print(pks)
```

### 여행 계획 세우기 객체

```python
# 여행 계획 세우기 요청 더미 객체
# 실제 시스템에선 모든 도메인의 상세 정보들이 travel_context에 들어있어야 함.
# 여행계획을 세우기 전 마지막 요청 사항을 질문에 답변을 통해 받는다고 가정 -> user_request
# 마지막 여행 계획을 세울까요? 같은 질문에 대한 사용자의 답변이 user_input으로 들어온다고 가정
generate_plan_request = {
    "user_input" : "당일치기 여행이라 숙소는 필요없어.",
    "travel_context" : [region_info1, region_info2, region_info3, place_info1, place_info2],
    "user_request" : "맛집의 리스트를 많이 뽑아줘",
    "prompt" : """
    당신은 전문적인 여행 플래너입니다. 사용자의 요청과 제공된 여행 정보를 바탕으로 구체적이고 실용적인 여행 계획을 작성해주세요.

    사용자 요청 : {user_request}

    여행 관련 정보 : {travel_context}

    여행 계획 작성 가이드라인 : 
    1. 사용자의 요청사항을 정확히 반영해주세요
    2. 제공된 여행 정보를 적극 활용해주세요
    3. 구체적인 장소명, 주소, 특징을 포함해주세요
    4. 시간대별 일정을 제안해주세요
    5. 교통편, 소요시간, 예상 비용 등 실용적인 정보를 포함해주세요
    6. 한국어로 친근하고 자연스럽게 작성해주세요

    여행 계획 :
    """
    
}
```

## 여행 계획 세우기

```python
# 프롬프트 템플릿 객체 생성
travel_plan_template = ChatPromptTemplate.from_template(generate_plan_request["prompt"])

# 템플릿에 내용을 넣어 메세지 구성
messages = travel_plan_template.format_messages(
    user_input = generate_plan_request["user_input"],
    user_request = generate_plan_request["user_request"],
    travel_context = generate_plan_request["travel_context"]
)

# llm 설정, temperature는 창의성을 위해 0.5로 설정
llm = ChatOpenAI(
    model= "gpt-4o-mini",
    temperature= 0.5
)

# 여행 플래너를 llm에게 지시
response = llm.invoke(messages)

# 결과 확인
print(response.content)
```
